```python
#Importing required libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
#Loading the dataset
df = pd.read_csv("DATASET/churn-bigml-80.csv")
```

```python
#checking basic details
print("First 5 rows:")
print(df.head())

print("\nColumn Info:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
#Separating numerical and categorical columns
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(include='object').columns
```

```python
#Filling missing numericals values with mean
for col in num_cols:
   df[col].fillna(df[col].mean(), inplace=True)
    
```

```python
#Filling missing numericals values with most frequent value
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
# Encode binary columns
df['international plan'] = df['international plan'].map({'Yes':1, 'No':0})
df['voice mail plan'] = df['voice mail plan'].map({'Yes':1, 'No':0})
df['churn'] = df['churn'].map({'True':1, 'False':0})
```

```python
# Clean column names first
df.columns = df.columns.str.strip()

# Check if state exists
if 'State' in df.columns:
    df = pd.get_dummies(df, columns=['State'], drop_first=True)
elif 'state' in df.columns:
    df = pd.get_dummies(df, columns=['state'], drop_first=True)
else:
    print("State column not found! Available columns:")
    print(df.columns.tolist())
```

```python
#Defining X and y
X = df.drop('churn', axis=1)
y = df['churn']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```
